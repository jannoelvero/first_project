#### CLEANING PLAN:

```python
import pandas as pd
import numpy as np
import re

# Load the dataset
df = pd.read_csv("../data/raw/airnb.csv")

# Keeping a copy
df_raw = df.copy()

# Inspect dimensions and first rows
print(df.shape)
display(df.head())
```

#### Standardize column names

```python
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
      .str.replace(r"[^\w]+", "_", regex=True)
      .str.strip("_")
)

print(df.columns.tolist())
```

#### Check the dataset structure

```python
print("Rows and columns:", df.shape)
print("\nData types:")
display(df.dtypes)

print("\nMissing values:")
display(df.isna().sum())

print("\nDuplicated rows:", df.duplicated().sum())
```

#### Removing irrelevant columns "offer_price" and "number of bed"

```python
columns_to_drop = ["offer_price", "number_of_bed"]

df = df.drop(columns=columns_to_drop, errors="ignore")

display(df.head())
print(df.columns.tolist())
```

#### Clean missing values and text formatting

```python
# Replace empty strings and whitespace-only values with NaN
df = df.replace(r"^\s*$", np.nan, regex=True)

# Clean all remaining text columns
text_columns = df.select_dtypes(include="object").columns

for column in text_columns:
    df[column] = df[column].str.strip()
```

```python
display(df.isna().sum())
```

#### Removing duplicate

```python
duplicate_rows = df[df.duplicated(keep=False)].sort_values(by="title")

display(duplicate_rows.head(10))
print("Rows involved in duplicates:", len(duplicate_rows))
```

```python
#df = df.drop_duplicates(keep="first").reset_index(drop=True)

print("Shape after removing duplicates:", df.shape)
print("Remaining duplicated rows:", df.duplicated().sum())
```

#### Split  review_and_rating 

```python
review_rating_text = df["review_and_rating"].fillna("").str.lower()
```

```python
df["rating"] = pd.to_numeric(
    review_rating_text.str.extract(
        r"(?P<rating>\d+(?:[.,]\d+)?)\s*(?:stars?|rating)?",
        expand=False
    ).str.replace(",", ".", regex=False),
    errors="coerce"
)
```

```python
df["review"] = pd.to_numeric(
    review_rating_text.str.extract(
        r"(?P<review>\d[\d,]*)\s*(?:reviews?|review)",
        expand=False
    ).str.replace(",", "", regex=False),
    errors="coerce"
)
```

```python
display(df[["review_and_rating", "rating", "review"]].head(10))
```

```python
print(df["rating"].describe())

invalid_ratings = df[
    (df["rating"].notna()) &
    ((df["rating"] < 0) | (df["rating"] > 5))
]

display(invalid_ratings)
```

#### Split  date  into day and month

```python
display(df["date"].dropna().head(10))
```

```python
parsed_date = pd.to_datetime(df["date"], errors="coerce")

df["day"] = parsed_date.dt.day
df["month"] = parsed_date.dt.month
```

```python
date_parts = df["date"].str.extract(
    r"(?i)(?P<month_name>[A-Za-z]+)\s+(?P<day>\d{1,2})"
)

df["day"] = pd.to_numeric(date_parts["day"], errors="coerce")
df["month_name"] = date_parts["month_name"].str.title()
```

```python
print("Unparsed dates:", parsed_date.isna().sum())
display(df.loc[parsed_date.isna(), ["date"]].head())
```

#### Convert  price  into a numeric column

```python
df["price_clean"] = (
    df["price_in_dollar"]
      .astype("string")
      .str.replace(r"[^\d,.\-]", "", regex=True)
      .str.replace(",", "", regex=False)
)

df["price_in_dollar"] = pd.to_numeric(df["price_clean"], errors="coerce")

df = df.drop(columns="price_clean")
```

```python
display(df[["price_in_dollar"]].head())
print(df["price_in_dollar"].describe())
print("Missing prices:", df["price_in_dollar"].isna().sum())
```

#### Split  title  into property type and city

```python
title_parts = df["title"].str.extract(
    r"(?i)^(?P<property_type>.+?)\s+in\s+(?P<city>.+?)\s*$"
)

df["property_type"] = title_parts["property_type"].str.strip()
df["city"] = title_parts["city"].str.strip()
df["country"] = "US"
```

```python
display(
    df[["title", "property_type", "city", "country"]].head(10)
)
```

```python
#Handling titles with different patterns
title_not_parsed = df[
    df["property_type"].isna() | df["city"].isna()
]

display(title_not_parsed[["title"]].head(20))
print("Titles not parsed:", len(title_not_parsed))
```

```python
title_parts = df["title"].str.extract(
    r"(?i)^(?P<property_type>.+?)\s+in\s+"
    r"(?P<city>[^,]+)"
    r"(?:,\s*(?P<country>[A-Za-z]{2}))?\s*$"
)

df["property_type"] = title_parts["property_type"].str.strip()
df["city"] = title_parts["city"].str.strip()
df["country"] = title_parts["country"].fillna("US").str.upper()
```

#### Create an amenities column from  detail 

```python
amenity_dictionary = {
    "pool": "Pool",
    "jacuzzi": "Jacuzzi",
    "hot tub": "Jacuzzi",
    "ocean view": "Ocean view",
    "sea view": "Ocean view",
    "ocean front": "Ocean front",
    "beachfront": "Ocean front",
    "Sauna": "sauna",
    "Nature": "nature",
    "Kayak": "kayak" , 
    "Waterfront": "waterfront" ,
    "Fireplace": "fireplace",
    "FREE Parking": "free parking" ,
    "wifi": "Wi-Fi",
    "wi-fi": "Wi-Fi",
    "kitchen": "Kitchen",
    "parking": "Parking",
    "air conditioning": "Air conditioning",
    "ac": "Air conditioning"
}
```

```python
def extract_amenities(text):
    if pd.isna(text):
        return np.nan

    text = str(text).lower()
    found = []

    for keyword, label in amenity_dictionary.items():
        if keyword in text and label not in found:
            found.append(label)

    return ", ".join(found) if found else np.nan
```

```python
df["amenities"] = df["detail"].apply(extract_amenities)

display(df[["detail", "amenities"]].head(10))
```
